# Exploración de Datos EEG

Este notebook contiene el análisis exploratorio de los datasets de MOABB


```python
import numpy as np
import matplotlib.pyplot as plt
from moabb.datasets import BNCI2014001
from moabb.paradigms import MotorImagery

# Cargar dataset
dataset = BNCI2014001()
paradigm = MotorImagery()

# Obtener datos para el primer sujeto
subjects = [1]
X, y, metadata = paradigm.get_data(dataset=dataset, subjects=subjects)

print(f"Forma de X: {X.shape}")
print(f"Forma de y: {y.shape}")
print(f"Clases únicas: {np.unique(y)}")
print(f"Número de canales: {X.shape[1]}")
print(f"Número de puntos temporales: {X.shape[2]}")
```

```python
# Visualizar una señal EEG de ejemplo
fig, axes = plt.subplots(4, 1, figsize=(12, 8))

for i, class_label in enumerate(np.unique(y)):
    # Encontrar el primer trial de esta clase
    idx = np.where(y == class_label)[0][0]
    
    # Plotear algunos canales
    for ch in range(min(5, X.shape[1])):
        axes[i].plot(X[idx, ch, :], label=f'Canal {ch}', alpha=0.7)
    
    axes[i].set_title(f'Clase: {class_label}')
    axes[i].set_xlabel('Tiempo (muestras)')
    axes[i].set_ylabel('Amplitud (uV)')
    axes[i].legend()
    axes[i].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Análisis estadístico básico
print("Estadísticas por clase:")
for class_label in np.unique(y):
    class_data = X[y == class_label]
    print(f"\nClase {class_label}:")
    print(f"  Número de trials: {len(class_data)}")
    print(f"  Media: {np.mean(class_data):.4f}")
    print(f"  Desviación estándar: {np.std(class_data):.4f}")
    print(f"  Min: {np.min(class_data):.4f}")
    print(f"  Max: {np.max(class_data):.4f}")
```
