# Visualización de Resultados

Este notebook contiene gráficos y análisis de los resultados del benchmark


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os

# Configurar estilo
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
# Cargar resultados del benchmark
results_dir = '../results'
results_file = os.path.join(results_dir, 'results.csv')

if os.path.exists(results_file):
    df = pd.read_csv(results_file)
    print("Resultados cargados:")
    print(df.head())
else:
    print("No se encontró el archivo de resultados. Ejecuta primero benchmark.py")
```

```python
# Gráfico de comparación de modelos
if 'df' in locals() and not df.empty:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Boxplot de accuracy por modelo
    if 'model' in df.columns and 'score' in df.columns:
        sns.boxplot(data=df, x='model', y='score', ax=axes[0])
        axes[0].set_title('Distribución de Accuracy por Modelo')
        axes[0].set_ylabel('Accuracy')
        axes[0].tick_params(axis='x', rotation=45)
    
    # Gráfico de accuracy por sujeto
    if 'subject' in df.columns:
        sns.lineplot(data=df, x='subject', y='score', hue='model', marker='o', ax=axes[1])
        axes[1].set_title('Accuracy por Sujeto')
        axes[1].set_xlabel('Sujeto')
        axes[1].set_ylabel('Accuracy')
        axes[1].legend(title='Modelo')
    
    plt.tight_layout()
    plt.savefig(os.path.join(results_dir, 'comparison_plot.png'), dpi=300, bbox_inches='tight')
    plt.show()
else:
    print("No hay datos para graficar")
```

```python
# Tabla resumen de resultados
if 'df' in locals() and not df.empty:
    if 'model' in df.columns and 'score' in df.columns:
        summary = df.groupby('model')['score'].agg(['mean', 'std', 'min', 'max'])
        summary.columns = ['Media', 'Desv. Est.', 'Mínimo', 'Máximo']
        summary = summary.round(4)
        print("\nResumen de Resultados:")
        print(summary)
```
